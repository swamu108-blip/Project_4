```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import io
```

```python
url = "https://en.wikipedia.org/wiki/Nationwide_opinion_polling_for_the_2016_United_States_presidential_election"
response = requests.get(url)
soup = BeautifulSoup(response.text)

response.status_code
```

```python
tables = soup.find_all("table", class_="wikitable")
```
